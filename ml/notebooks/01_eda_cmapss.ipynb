```python
import sys
sys.path.append('../../')

import pandas as pd
import matplotlib.pyplot as plt

from ml.data.load_cmapss import load_cmapss
from ml.data.load_ai4i import load_ai4i

cmapss_train, cmapss_test, cmapss_rul = load_cmapss()
ai4i = load_ai4i()

print("CMAPSS train:", cmapss_train.shape)
print("CMAPSS test:", cmapss_test.shape)
print("RUL:", cmapss_rul.shape)
print("AI4I:", ai4i.shape)
```

```python
print("CMAPSS TRAIN HEAD")
display(cmapss_train.head())

print("\nCMAPSS TRAIN INFO")
cmapss_train.info()

print("\nCMAPSS MISSING VALUES")
print(cmapss_train.isnull().sum().sum(), "missing values in train")

print("\nAI4I HEAD")
display(ai4i.head())

print("\nAI4I INFO")
ai4i.info()

print("\nAI4I MISSING VALUES")
print(ai4i.isnull().sum().sum(), "missing values")
```

```python
print("CMAPSS DESCRIBE")
display(cmapss_train.describe())

print("\nAI4I DESCRIBE")
display(ai4i.describe())

print("\nAI4I FAILURE RATE")
print(ai4i['Machine failure'].value_counts())
print(ai4i['Machine failure'].value_counts(normalize=True))
```

```python
# Pick one engine and plot a few sensors over its lifetime
engine_1 = cmapss_train[cmapss_train['unit_number'] == 1]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
sensors_to_plot = ['sensor_measurement_2', 'sensor_measurement_7', 'sensor_measurement_11', 'sensor_measurement_15']

for ax, sensor in zip(axes.flat, sensors_to_plot):
    ax.plot(engine_1['time_in_cycles'], engine_1[sensor])
    ax.set_title(sensor)
    ax.set_xlabel('cycle')

plt.tight_layout()
plt.show()
```

```python
ai4i['Machine failure'].value_counts().plot(kind='bar', title='AI4I Failure vs No Failure')
plt.show()
```

```python
# Identify truly zero/near-zero variance columns across BOTH train and test
threshold = 1e-6  # anything below this std is considered "effectively constant"

train_std = cmapss_train.drop(columns=['unit_number', 'time_in_cycles']).std()
dead_cols = train_std[train_std < threshold].index.tolist()

print("Columns with near-zero variance in TRAIN:")
print(dead_cols)

# Sanity-check: confirm these are ALSO dead in test
test_std = cmapss_test.drop(columns=['unit_number', 'time_in_cycles']).std()
print("\nSame columns' std in TEST:")
print(test_std[dead_cols])
```

```python
from ml.data.clean_cmapss import clean_cmapss

cleaned_train, cleaned_test = clean_cmapss(cmapss_train, cmapss_test)

print("Before cleaning:", cmapss_train.shape)
print("After cleaning:", cleaned_train.shape)
print("\nRemaining columns:")
print(cleaned_train.columns.tolist())
```

```python
# Check how strongly the 5 subtype flags predict Machine failure
leak_cols = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']

# If ANY flag is 1, is Machine failure always 1?
any_flag_set = ai4i[leak_cols].any(axis=1)
print("Rows where any flag=1 AND Machine failure=1:", ((any_flag_set) & (ai4i['Machine failure']==1)).sum())
print("Rows where any flag=1 AND Machine failure=0:", ((any_flag_set) & (ai4i['Machine failure']==0)).sum())
print("Rows where no flag set AND Machine failure=1:", ((~any_flag_set) & (ai4i['Machine failure']==1)).sum())
print("Rows where no flag set AND Machine failure=0:", ((~any_flag_set) & (ai4i['Machine failure']==0)).sum())
```

```python
from ml.data.clean_ai4i import clean_ai4i

cleaned_ai4i = clean_ai4i(ai4i)
print("Before:", ai4i.shape)
print("After:", cleaned_ai4i.shape)
print("\nRemaining columns:")
print(cleaned_ai4i.columns.tolist())
print("\nType column values:", cleaned_ai4i['Type'].unique())
```

```python
from ml.features.cmapss_features import build_cmapss_features

train_feat, test_feat, meta = build_cmapss_features(cleaned_train, cleaned_test, cmapss_rul)
print("Train RUL range:", train_feat['rul'].min(), "-", train_feat['rul'].max())
print("Test RUL range:", test_feat['rul'].min(), "-", test_feat['rul'].max())
print("Any NaN in train rul:", train_feat['rul'].isna().sum())
print("Any NaN in test rul:", test_feat['rul'].isna().sum())
```

```python
from ml.features.ai4i_features import build_ai4i_features

scaled_ai4i, failure_type, ai4i_meta = build_ai4i_features(cleaned_ai4i, ai4i)

print("Scaled shape:", scaled_ai4i.shape)
print("New feature columns:", [c for c in scaled_ai4i.columns if c not in cleaned_ai4i.columns])
print("\nAny NaN in scaled features:", scaled_ai4i.isna().sum().sum())

print("\nfailure_type value counts:")
print(failure_type.value_counts())

print("\nDoes failure_type ever appear inside scaled_ai4i's columns?", 'failure_type' in scaled_ai4i.columns)
```

```python
from ml.models.classification import train as train_classification

model, multi_model, metrics = train_classification(scaled_ai4i.assign(**{
    'Machine_failure': cleaned_ai4i['Machine_failure'].values,
    'failure_type': failure_type.values
}))

import json
print(json.dumps(metrics, indent=2))
```

```python
from ml.models.anomaly_isolation_forest import train as train_anomaly

anomaly_model, anomaly_metrics = train_anomaly(scaled_ai4i.assign(**{
    'Machine_failure': cleaned_ai4i['Machine_failure'].values,
    'failure_type': failure_type.values
}))
print(anomaly_metrics)
```

```python
from ml.models.anomaly_autoencoder import train as train_autoencoder

ae_model, ae_threshold, ae_metrics = train_autoencoder(scaled_ai4i.assign(**{
    'Machine_failure': cleaned_ai4i['Machine_failure'].values,
    'failure_type': failure_type.values
}))
print(ae_metrics)
```

```python
from ml.models.regression import train as train_regression

train_feat, test_feat, cmapss_meta = build_cmapss_features(cleaned_train, cleaned_test, cmapss_rul)
reg_model, reg_metrics = train_regression(train_feat, test_feat)
print(reg_metrics)
```

```python
print("Train columns:", train_feat.drop(columns=['rul']).columns.tolist())
print("\nTest columns:", test_feat.drop(columns=['rul']).columns.tolist())
print("\nSame columns, same order?", 
      train_feat.drop(columns=['rul']).columns.tolist() == test_feat.drop(columns=['rul']).columns.tolist())

print("\nTrain feature NaN count:", train_feat.drop(columns=['rul']).isna().sum().sum())
print("Test feature NaN count:", test_feat.drop(columns=['rul']).isna().sum().sum())

print("\nTrain rows per engine (min/max):", train_feat.groupby('unit_number').size().min(), "/", train_feat.groupby('unit_number').size().max())
print("Test rows per engine (min/max):", test_feat.groupby('unit_number').size().min(), "/", test_feat.groupby('unit_number').size().max())
```

```python
print("Train RUL distribution:")
print(train_feat['rul'].describe())
print("\nTest RUL distribution:")
print(test_feat['rul'].describe())

# What does the model actually predict, compared to the true test RUL?
import numpy as np
X_test_only = test_feat.drop(columns=['rul', 'unit_number', 'time_in_cycles'])
preds = reg_model.predict(X_test_only)
print("\nPredicted RUL - min/mean/max:", preds.min(), preds.mean(), preds.max())
print("True test RUL - min/mean/max:", test_feat['rul'].min(), test_feat['rul'].mean(), test_feat['rul'].max())
```

```python
print("train_feat index is monotonic/sequential:", train_feat.index.is_monotonic_increasing)
print("train_feat index has duplicates:", train_feat.index.duplicated().sum())
print("train_feat index min/max/len:", train_feat.index.min(), train_feat.index.max(), len(train_feat))

# Sanity check on one specific engine, does row order within it make physical sense?
engine1 = train_feat[train_feat['unit_number'] == 1].sort_index()
print("\nEngine 1 rul values, in index order (should DECREASE steadily to 0):")
print(engine1['rul'].head(10).tolist())
print(engine1['rul'].tail(10).tolist())
```

```python
# Compare a few key sensors' distributions between train and test
key_sensors = ['sensor_measurement_11', 'sensor_measurement_4', 'sensor_measurement_15']

for sensor in key_sensors:
    print(f"=== {sensor} ===")
    print("Train:", train_feat[sensor].describe()[['mean', 'std', 'min', 'max']].to_dict())
    print("Test: ", test_feat[sensor].describe()[['mean', 'std', 'min', 'max']].to_dict())
    print()

# The real test: does a LOW true RUL actually correspond to visibly
# different sensor readings in test, the way it does in train?
print("=== In TRAIN: sensor_measurement_11 at low RUL (near failure) vs high RUL (healthy) ===")
print("RUL < 10:", train_feat[train_feat['rul'] < 10]['sensor_measurement_11'].mean())
print("RUL > 100:", train_feat[train_feat['rul'] > 100]['sensor_measurement_11'].mean())

print("\n=== In TEST: sensor_measurement_11 at low RUL vs high RUL ===")
print("RUL < 10:", test_feat[test_feat['rul'] < 10]['sensor_measurement_11'].mean())
print("RUL > 100:", test_feat[test_feat['rul'] > 100]['sensor_measurement_11'].mean())
```

```python
from sklearn.model_selection import GroupShuffleSplit
from ml.models.lstm_rul import train as train_lstm

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
groups = train_feat['unit_number']
train_idx, val_idx = next(gss.split(train_feat, groups=groups))
train_split = train_feat.iloc[train_idx]
val_split = train_feat.iloc[val_idx]

lstm_model_val, lstm_metrics_val = train_lstm(train_split, val_split)
print("HELD-OUT TRAIN-ENGINE VALIDATION:", lstm_metrics_val)
```
